```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_parquet(
    r"D:\Guvi\Projects\ride_flow_ai\data\raw\yellow_tripdata_2024-01.parquet"
)

print(df.head())
```

```python
df = df.head(100000)
```

```python
print(df.shape)

print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
df['tpep_pickup_datetime'] = pd.to_datetime(
    df['tpep_pickup_datetime']
)

df['tpep_dropoff_datetime'] = pd.to_datetime(
    df['tpep_dropoff_datetime']
)
```

```python
df['pickup_hour'] = (
    df['tpep_pickup_datetime']
    .dt.hour
)

df['pickup_day'] = (
    df['tpep_pickup_datetime']
    .dt.day
)

df['pickup_weekday'] = (
    df['tpep_pickup_datetime']
    .dt.weekday
)

df['pickup_month'] = (
    df['tpep_pickup_datetime']
    .dt.month
)
```

```python
df['is_weekend'] = (
    df['pickup_weekday']
    .isin([5,6])
    .astype(int)
)
```

```python
df['trip_duration'] = (
    df['tpep_dropoff_datetime']
    -
    df['tpep_pickup_datetime']
).dt.total_seconds() / 60    
```

```python
df = df[df['trip_duration'] > 0]
```

```python
print(df.head())

print(df.shape)
```

```python
df.to_csv(
    r"D:\Guvi\Projects\ride_flow_ai\data\preprocessed/cleaned_rides.csv",
    index=False
)

print("Dataset Saved Successfully")
```
